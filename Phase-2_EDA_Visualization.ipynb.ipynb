# 🔶 Univariate Analaysis

## Disaster Type Distribution

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='disaster_type')
plt.title("Distribution of Disaster Types")
plt.xticks(rotation=45)
plt.show()
```

## Severity Index Distribution

```python
plt.figure(figsize=(7,5))
sns.histplot(df['severity_index'], bins=10, kde=True)
plt.title("Severity Index Distribution")
plt.show()
```

## Country Distribution

```python
top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(8,5))
top_countries.plot(kind='bar')
plt.title("Top 10 Countries by Disaster Count")
plt.ylabel("Number of Disasters")
plt.show()
```

## Country-wise Disaster Distribution

```python
country_counts = df['country'].value_counts()

plt.figure(figsize=(7,7))
plt.pie(
    country_counts,
    labels=country_counts.index,
    autopct='%1.1f%%',
    startangle=140
)
plt.title("Country-wise Disaster Distribution")
plt.show()
```

- Countries with slightly higher shares indicate greater disaster exposure or reporting frequency.
- The distribution highlights the global nature of disasters, emphasizing the need for international preparedness and cooperation.

## Numerical vs Numerical (Num–Num) Analysis

```python
plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x='severity_index', y='casualties')
plt.title("Severity Index vs Casualties")
plt.xlabel("Severity Index")
plt.ylabel("Casualties")
plt.show()
```

- This scatter plot shows the relationship between disaster severity and human impact.

- Each dot represents one disaster event.

## Numerical vs Numerical (Num–Num) Analysis

```python
plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x='response_time_hours', y='recovery_days')
plt.title("Response Time vs Recovery Days")
plt.xlabel("Response Time (Hours)")
plt.ylabel("Recovery Days")
plt.show()
```

- This scatter plot shows the relationship between emergency response time and recovery duration.

## Categorical vs Numerical (Cat–Num) Analysis

```python
plt.figure(figsize=(8,5))
df.groupby('disaster_type')['severity_index'].mean().plot(kind='bar')
plt.title("Disaster Type vs Average Severity Index")
plt.xlabel("Disaster Type")
plt.ylabel("Average Severity Index")
plt.xticks(rotation=45)
plt.show()
```

- The chart shows how average disaster severity varies across different disaster types.
- Certain disaster types exhibit higher average severity, indicating more intense or damaging events.
- Other disaster categories show relatively moderate severity levels, suggesting lower average intensity.
- The comparison helps identify high-severity disaster types, which require stronger preparedness and mitigation measures.

## Categorical vs Categorical Analysis

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.countplot(data=df, x='country', hue='disaster_type')
plt.title("Country vs Disaster Type")
plt.xlabel("Country")
plt.ylabel("Number of Disasters")
plt.xticks(rotation=45)
plt.show()
```

- The chart shows how different disaster types are distributed across countries.

- The distribution highlights the importance of country-specific disaster preparedness and response strategies.

## Multivaraiate Anyalsis

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x='severity_index',
    y='economic_loss_usd',
    hue='disaster_type'
)
plt.title("Severity Index vs Economic Loss by Disaster Type")
plt.xlabel("Severity Index")
plt.ylabel("Economic Loss (USD)")
plt.show()
```

- The plot shows a positive relationship between disaster severity and economic loss across all disaster types.

- Higher severity events generally result in greater economic damage, regardless of disaster category.

- Different disaster types follow a similar upward trend, indicating that severity is a major driver of economic impact.

- The spread of points at higher severity levels suggests variation in losses, influenced by disaster type, location, and response effectiveness.

**Insight:** Higher severity disasters tend to cause greater economic loss across all disaster types.

## Time Based Anyalsis

```python
df['year'] = pd.to_datetime(df['date']).dt.year
```

```python
df['year'] = df['date']
plt.figure(figsize=(8,5))
df.groupby('year').size().plot(kind='line', marker='o')
plt.title("Number of Disasters Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Disasters")
plt.show()
```

- The line chart shows year-wise changes in the number of disaster events over the study period.
- Disaster occurrences fluctuate across years, indicating variability in annual disaster activity.
- Certain years exhibit higher disaster counts, highlighting periods of increased disaster intensity.
- The overall trend emphasizes the need for continuous monitoring and long-term disaster preparedness.

## Correlation heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt

# select numerical columns & calculate correlation
corr_matrix = df.select_dtypes(include='number').corr()

# plot heatmap
plt.figure(figsize=(8,6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

- Severity index shows strong positive correlations with economic loss, casualties, and recovery days, indicating higher severity leads to greater impact.
- Response time is negatively correlated with response efficiency score, showing that quicker responses improve efficiency.
- Aid amount has a positive relationship with severity and recovery duration, suggesting higher aid allocation for severe disasters.
- Geographical variables (latitude and longitude) show weak or no correlation with impact and response metrics.

**Insight:** Severity index and response time are the strongest drivers influencing disaster impact and recovery.
