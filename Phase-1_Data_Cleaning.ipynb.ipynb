## → What kind of data does each column contain?

- 1.date:
Contains the date on which the disaster occurred.

- 2.country:
Represents the country where the disaster took place.

- 3.disaster_type:
Indicates the type of disaster (such as flood, earthquake, cyclone, etc.).

- 4.severity_index:
A numerical measure representing the intensity or severity of the disaster.

- 5.casualties:
The total number of people injured or killed due to the disaster.

- 6.economic_loss_usd:
The estimated financial loss caused by the disaster, measured in US dollars.

- 7.response_time_hours:
The number of hours taken for emergency response teams to reach the affected area.

- 8.aid_amount_usd:
The amount of financial aid provided for disaster relief, in US dollars.

- 9.response_efficiency_score:
A numerical score that represents how effective the disaster response was.

- 10.recovery_days:
The number of days required for the affected region to return to normal conditions.

- 11.latitude:
The geographical latitude of the disaster location.

- 12.longitude:
The geographical longitude of the disaster location.

### → Are there any missing or unusual values?

- All columns follow appropriate data types (dates, numerical values, and categorical values).

- No unusual or inconsistent values were observed, such as negative losses or invalid response times.
